```python
# ============================================================
# RESOLVEAI
# 01 - Exploratory Data Analysis
# ============================================================


# ============================================================
# CELL 1 — IMPORT LIBRARIES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)

print("Libraries imported successfully.")


# ============================================================
# CELL 2 — LOAD DATASET
# ============================================================

# Notebook is inside:
# ResolveAI/notebooks/
#
# Dataset is inside:
# ResolveAI/data/raw/tickets.csv

DATA_PATH = Path("../data/raw/tickets.csv")

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")


# ============================================================
# CELL 3 — FIRST LOOK
# ============================================================

print("\n" + "=" * 70)
print("FIRST 5 ROWS")
print("=" * 70)

display(df.head())


print("\n" + "=" * 70)
print("RANDOM 5 ROWS")
print("=" * 70)

display(df.sample(5, random_state=42))


# ============================================================
# CELL 4 — COLUMN INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("COLUMN INFORMATION")
print("=" * 70)

column_info = pd.DataFrame({
    "column": df.columns,
    "dtype": df.dtypes.astype(str),
    "missing_count": df.isna().sum().values,
    "missing_%": (df.isna().mean() * 100).round(2).values,
    "unique_values": df.nunique().values
})

display(column_info)


# ============================================================
# CELL 5 — DATASET INFO
# ============================================================

print("\n" + "=" * 70)
print("DATASET INFO")
print("=" * 70)

df.info()


# ============================================================
# CELL 6 — DUPLICATE ANALYSIS
# ============================================================

print("\n" + "=" * 70)
print("DUPLICATE ANALYSIS")
print("=" * 70)

full_duplicates = df.duplicated().sum()

text_duplicates = df.duplicated(
    subset=["subject", "body"]
).sum()

print(f"Completely duplicated rows : {full_duplicates:,}")
print(f"Duplicate subject + body   : {text_duplicates:,}")


# ============================================================
# CELL 7 — UNIQUE VALUES OF IMPORTANT COLUMNS
# ============================================================

important_columns = [
    "language",
    "type",
    "queue",
    "priority",
    "version"
]

for column in important_columns:

    print("\n" + "=" * 70)
    print(f"UNIQUE VALUES — {column.upper()}")
    print("=" * 70)

    print(df[column].unique())


# ============================================================
# CELL 8 — LANGUAGE DISTRIBUTION
# ============================================================

language_counts = df["language"].value_counts()

print("\n" + "=" * 70)
print("LANGUAGE DISTRIBUTION")
print("=" * 70)

display(language_counts.to_frame("count"))

print("\nPercentage:")
display(
    (df["language"].value_counts(normalize=True) * 100)
    .round(2)
    .to_frame("percentage")
)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="language",
    order=language_counts.index
)

plt.title("Ticket Distribution by Language")
plt.xlabel("Language")
plt.ylabel("Number of Tickets")
plt.tight_layout()
plt.show()


# ============================================================
# CELL 9 — TICKET TYPE DISTRIBUTION
# ============================================================

type_counts = df["type"].value_counts()

print("\n" + "=" * 70)
print("TICKET TYPE DISTRIBUTION")
print("=" * 70)

display(type_counts.to_frame("count"))


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="type",
    order=type_counts.index
)

plt.title("Ticket Distribution by Type")
plt.xlabel("Ticket Type")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# CELL 10 — SUPPORT QUEUE DISTRIBUTION
# ============================================================

queue_counts = df["queue"].value_counts()

print("\n" + "=" * 70)
print("SUPPORT QUEUE DISTRIBUTION")
print("=" * 70)

display(queue_counts.to_frame("count"))

print("\nPercentage:")
display(
    (df["queue"].value_counts(normalize=True) * 100)
    .round(2)
    .to_frame("percentage")
)


plt.figure(figsize=(11, 6))

sns.countplot(
    data=df,
    y="queue",
    order=queue_counts.index
)

plt.title("Ticket Distribution by Support Queue")
plt.xlabel("Number of Tickets")
plt.ylabel("Support Queue")
plt.tight_layout()
plt.show()


# ============================================================
# CELL 11 — PRIORITY DISTRIBUTION
# ============================================================

priority_counts = df["priority"].value_counts()

print("\n" + "=" * 70)
print("PRIORITY DISTRIBUTION")
print("=" * 70)

display(priority_counts.to_frame("count"))

print("\nPercentage:")
display(
    (df["priority"].value_counts(normalize=True) * 100)
    .round(2)
    .to_frame("percentage")
)


plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="priority",
    order=priority_counts.index
)

plt.title("Ticket Distribution by Priority")
plt.xlabel("Priority")
plt.ylabel("Number of Tickets")
plt.tight_layout()
plt.show()


# ============================================================
# CELL 12 — TEXT PREPARATION FOR EDA
# ============================================================

# IMPORTANT:
# We are NOT permanently cleaning the dataset here.
# These are temporary columns used only for EDA.

df["subject_eda"] = df["subject"].fillna("").astype(str)
df["body_eda"] = df["body"].fillna("").astype(str)

df["combined_text"] = (
    df["subject_eda"].str.strip()
    + " "
    + df["body_eda"].str.strip()
).str.strip()

df["word_count"] = (
    df["combined_text"]
    .str.split()
    .str.len()
)

df["character_count"] = (
    df["combined_text"]
    .str.len()
)

print("Text analysis columns created.")


# ============================================================
# CELL 13 — TEXT STATISTICS
# ============================================================

print("\n" + "=" * 70)
print("TEXT STATISTICS")
print("=" * 70)

text_statistics = df[
    ["word_count", "character_count"]
].describe()

display(text_statistics)


# ============================================================
# CELL 14 — WORD COUNT DISTRIBUTION
# ============================================================

plt.figure(figsize=(10, 5))

sns.histplot(
    df["word_count"],
    bins=50,
    kde=True
)

plt.title("Distribution of Ticket Word Counts")
plt.xlabel("Number of Words")
plt.ylabel("Number of Tickets")
plt.tight_layout()
plt.show()


# ============================================================
# CELL 15 — TICKET LENGTH BY LANGUAGE
# ============================================================

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="language",
    y="word_count"
)

plt.title("Ticket Length by Language")
plt.xlabel("Language")
plt.ylabel("Word Count")
plt.tight_layout()
plt.show()


# ============================================================
# CELL 16 — TICKET LENGTH BY TYPE
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="type",
    y="word_count"
)

plt.title("Ticket Length by Ticket Type")
plt.xlabel("Ticket Type")
plt.ylabel("Word Count")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()


# ============================================================
# CELL 17 — LANGUAGE VS TYPE
# ============================================================

language_type = pd.crosstab(
    df["language"],
    df["type"]
)

print("\n" + "=" * 70)
print("LANGUAGE VS TICKET TYPE")
print("=" * 70)

display(language_type)


plt.figure(figsize=(9, 5))

sns.heatmap(
    language_type,
    annot=True,
    fmt="d"
)

plt.title("Language vs Ticket Type")
plt.xlabel("Ticket Type")
plt.ylabel("Language")
plt.tight_layout()
plt.show()


# ============================================================
# CELL 18 — LANGUAGE VS PRIORITY
# ============================================================

language_priority = pd.crosstab(
    df["language"],
    df["priority"]
)

print("\n" + "=" * 70)
print("LANGUAGE VS PRIORITY")
print("=" * 70)

display(language_priority)


plt.figure(figsize=(9, 5))

sns.heatmap(
    language_priority,
    annot=True,
    fmt="d"
)

plt.title("Language vs Priority")
plt.xlabel("Priority")
plt.ylabel("Language")
plt.tight_layout()
plt.show()


# ============================================================
# CELL 19 — QUEUE VS PRIORITY
# ============================================================

queue_priority = pd.crosstab(
    df["queue"],
    df["priority"]
)

print("\n" + "=" * 70)
print("QUEUE VS PRIORITY")
print("=" * 70)

display(queue_priority)


plt.figure(figsize=(11, 7))

sns.heatmap(
    queue_priority,
    annot=True,
    fmt="d"
)

plt.title("Support Queue vs Priority")
plt.xlabel("Priority")
plt.ylabel("Support Queue")
plt.tight_layout()
plt.show()


# ============================================================
# CELL 20 — TYPE VS PRIORITY
# ============================================================

type_priority = pd.crosstab(
    df["type"],
    df["priority"]
)

print("\n" + "=" * 70)
print("TICKET TYPE VS PRIORITY")
print("=" * 70)

display(type_priority)


plt.figure(figsize=(9, 5))

sns.heatmap(
    type_priority,
    annot=True,
    fmt="d"
)

plt.title("Ticket Type vs Priority")
plt.xlabel("Priority")
plt.ylabel("Ticket Type")
plt.tight_layout()
plt.show()


# ============================================================
# CELL 21 — TAG ANALYSIS
# ============================================================

tag_columns = [
    "tag_1",
    "tag_2",
    "tag_3",
    "tag_4",
    "tag_5",
    "tag_6",
    "tag_7",
    "tag_8"
]

all_tags = df[tag_columns].stack().dropna()

tag_counts = all_tags.value_counts()

print("\n" + "=" * 70)
print("TOP 20 TAGS")
print("=" * 70)

display(tag_counts.head(20).to_frame("count"))


plt.figure(figsize=(10, 7))

sns.barplot(
    x=tag_counts.head(20).values,
    y=tag_counts.head(20).index
)

plt.title("Top 20 Ticket Tags")
plt.xlabel("Number of Occurrences")
plt.ylabel("Tag")
plt.tight_layout()
plt.show()


# ============================================================
# CELL 22 — MISSING SUBJECT ANALYSIS
# ============================================================

missing_subject = df["subject"].isna().sum()
missing_body = df["body"].isna().sum()

print("\n" + "=" * 70)
print("TEXT FIELD MISSING VALUES")
print("=" * 70)

print(f"Missing subjects : {missing_subject:,}")
print(f"Missing bodies   : {missing_body:,}")


# ============================================================
# CELL 23 — TICKETS WITH VERY LITTLE TEXT
# ============================================================

short_tickets = (
    df[df["word_count"] <= 5]
    [
        [
            "subject",
            "body",
            "type",
            "queue",
            "priority",
            "language",
            "word_count"
        ]
    ]
    .sort_values("word_count")
)

print("\n" + "=" * 70)
print("VERY SHORT TICKETS")
print("=" * 70)

print(f"Tickets with <= 5 words: {len(short_tickets):,}")

display(short_tickets.head(20))


# ============================================================
# CELL 24 — LONGEST TICKETS
# ============================================================

longest_tickets = (
    df.sort_values(
        "word_count",
        ascending=False
    )
    [
        [
            "subject",
            "body",
            "type",
            "queue",
            "priority",
            "language",
            "word_count"
        ]
    ]
)

print("\n" + "=" * 70)
print("LONGEST TICKETS")
print("=" * 70)

display(longest_tickets.head(10))


# ============================================================
# CELL 25 — ENGLISH DATASET
# ============================================================

english_df = df[
    df["language"].str.lower() == "en"
].copy()

print("\n" + "=" * 70)
print("ENGLISH DATASET")
print("=" * 70)

print(f"English tickets : {len(english_df):,}")
print(
    f"Percentage      : "
    f"{len(english_df) / len(df) * 100:.2f}%"
)


# ============================================================
# CELL 26 — ENGLISH TARGET DISTRIBUTIONS
# ============================================================

print("\n" + "=" * 70)
print("ENGLISH — QUEUE DISTRIBUTION")
print("=" * 70)

display(
    english_df["queue"]
    .value_counts()
    .to_frame("count")
)


print("\n" + "=" * 70)
print("ENGLISH — PRIORITY DISTRIBUTION")
print("=" * 70)

display(
    english_df["priority"]
    .value_counts()
    .to_frame("count")
)


print("\n" + "=" * 70)
print("ENGLISH — TYPE DISTRIBUTION")
print("=" * 70)

display(
    english_df["type"]
    .value_counts()
    .to_frame("count")
)


# ============================================================
# CELL 27 — SAMPLE ENGLISH TICKETS
# ============================================================

print("\n" + "=" * 70)
print("SAMPLE ENGLISH TICKETS")
print("=" * 70)

sample_columns = [
    "subject",
    "body",
    "type",
    "queue",
    "priority"
]

display(
    english_df[
        sample_columns
    ].sample(
        min(10, len(english_df)),
        random_state=42
    )
)


# ============================================================
# CELL 28 — CHECK TARGET MISSING VALUES
# ============================================================

target_columns = [
    "queue",
    "priority",
    "type"
]

print("\n" + "=" * 70)
print("TARGET VARIABLE QUALITY")
print("=" * 70)

target_quality = pd.DataFrame({
    "missing": df[target_columns].isna().sum(),
    "missing_%": (
        df[target_columns].isna().mean() * 100
    ).round(2),
    "unique_classes": df[target_columns].nunique()
})

display(target_quality)


# ============================================================
# CELL 29 — FINAL DATASET SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("RESOLVEAI — EDA SUMMARY")
print("=" * 70)

print(f"""
Total tickets              : {len(df):,}
Total columns              : {df.shape[1]}

Languages                  : {df['language'].nunique()}
English tickets            : {(df['language'] == 'en').sum():,}
German tickets             : {(df['language'] == 'de').sum():,}

Ticket types               : {df['type'].nunique()}
Support queues             : {df['queue'].nunique()}
Priority levels            : {df['priority'].nunique()}

Duplicate rows             : {df.duplicated().sum():,}
Duplicate subject + body   : {df.duplicated(['subject', 'body']).sum():,}

Missing subjects           : {df['subject'].isna().sum():,}
Missing bodies             : {df['body'].isna().sum():,}

Average words/ticket       : {df['word_count'].mean():.2f}
Median words/ticket        : {df['word_count'].median():.2f}
Maximum words/ticket       : {df['word_count'].max():,}
""")

print("=" * 70)
print("EDA COMPLETE")
print("=" * 70)
```
