```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)
```

```python
DATA_PATH = Path("../data/raw/tickets.csv")

df = pd.read_csv(DATA_PATH)

df.shape
```

```python
df.sample(5,random_state=42)
```

```python
duplicate_summary = pd.Series({
    "full_row_duplicates": df.duplicated().sum(),
    "subject_body_duplicates": df.duplicated(
        subset=["subject", "body"]
    ).sum()
})

duplicate_summary
```

```python
important_columns = [
    "language",
    "type",
    "queue",
    "priority",
    "version"
]

for column in important_columns:
    print(f"{column}:")
    print(df[column].unique())
    print()
```

```python
language_counts = df["language"].value_counts()

language_counts
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="language",
    order=language_counts.index
)

plt.title("Ticket Distribution by Language")
plt.xlabel("Language")
plt.ylabel("Number of Tickets")
plt.tight_layout()
plt.show()
```

```python
type_counts = df["type"].value_counts()

type_counts
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="type",
    order=type_counts.index
)

plt.title("Ticket Distribution by Type")
plt.xlabel("Ticket Type")
plt.ylabel("Number of Tickets")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
queue_counts = df["queue"].value_counts()

queue_counts
```

```python
plt.figure(figsize=(11, 6))

sns.countplot(
    data=df,
    y="queue",
    order=queue_counts.index
)

plt.title("Ticket Distribution by Support Queue")
plt.xlabel("Number of Tickets")
plt.ylabel("Support Queue")
plt.tight_layout()
plt.show()
```

```python
priority_counts = df["priority"].value_counts()

priority_counts
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="priority",
    order=priority_counts.index
)

plt.title("Ticket Distribution by Priority")
plt.xlabel("Priority")
plt.ylabel("Number of Tickets")
plt.tight_layout()
plt.show()
```

```python
df["subject_eda"] = df["subject"].fillna("").astype(str)
df["body_eda"] = df["body"].fillna("").astype(str)

df["combined_text"] = (
    df["subject_eda"].str.strip()
    + " "
    + df["body_eda"].str.strip()
).str.strip()

df["word_count"] = df["combined_text"].str.split().str.len()
df["character_count"] = df["combined_text"].str.len()
```

```python
df[["word_count", "character_count"]].describe()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    df["word_count"],
    bins=50,
    kde=True
)

plt.title("Distribution of Ticket Word Counts")
plt.xlabel("Number of Words")
plt.ylabel("Number of Tickets")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="language",
    y="word_count"
)

plt.title("Ticket Length by Language")
plt.xlabel("Language")
plt.ylabel("Word Count")
plt.tight_layout()
plt.show()
```
